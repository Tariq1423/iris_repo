```python
import pandas as pd

df = pd.read_csv('data/iris.csv')


display(df.head())


display(df.tail())


print("Shape:", df.shape)
print("Columns:", list(df.columns))
print("Data Types:\n", df.dtypes)
```

In this cell, we imported the pandas library and read the contents of the CSV file into a DataFrame named `df`.

```python

print("Missing values per column:\n", df.isnull().sum())


print("Duplicate rows count:", df.duplicated().sum())


print("Unique species count:", df['species'].nunique())
print("Flower count per species:\n", df['species'].value_counts())


print("Average Sepal Length:", round(df['sepal_length'].mean(), 2))
print("Maximum Petal Length:", df['petal_length'].max())
print("Minimum Sepal Width:", df['sepal_width'].min())

# Groupby mean calculation
species_summary = df.groupby('species').mean(numeric_only=True)
display(species_summary)


display(df.describe())
```

In this cell, we imported the pandas library as `pd` and loaded the Iris dataset from `data/iris.csv` into a DataFrame named `df`.

```python

filtered_sepal = df[df['sepal_length'] > 6]
filtered_petal = df[df['petal_width'] > 1.5]
setosa_df = df[df['species'] == 'setosa']

display(filtered_sepal.head())
display(filtered_petal.head())
display(setosa_df.head())


highest_petal_species = df.groupby('species')['petal_length'].mean().idxmax()
print("Species with highest average petal length:", highest_petal_species)


sorted_df = df.sort_values(by='sepal_length', ascending=False)
display(sorted_df.head())
```

# Data Filtering, Aggregation, and Sorting

In this section, we process and extract specific insights from the Iris dataset using boolean indexing, grouping, and sorting operations:

1. **Dataset Filtering (Boolean Indexing):**
   - **`filtered_sepal`**: Filters rows where `sepal_length` is greater than 6.0.
   - **`filtered_petal`**: Filters rows where `petal_width` is greater than 1.5.
   - **`setosa_df`**: Extracts all records belonging strictly to the `setosa` species.
   - We display the first 5 rows of each filtered subset using `.head()`.

2. **Group-Based Aggregation:**
   - We group the dataset by `species` and calculate the average `petal_length` for each group.
   - We use the `.idxmax()` function to return the specific species name (`virginica`) that holds the highest average petal length.

3. **Data Sorting:**
   - We sort the entire dataset based on `sepal_length` in descending order using `sort_values(by='sepal_length', ascending=False)`.
   - We display the top 5 rows of the sorted DataFrame.

```python
import os


os.makedirs('output', exist_ok=True)


species_summary.to_csv('output/species_summary.csv')
filtered_sepal.to_csv('output/filtered_iris.csv', index=False)

print("Files successfully exported to output directory.")
```

# Exporting Analysis Results to CSV

In this section, we save the summarized metrics and filtered data into output CSV files for future use and reporting:

1. **Creating Output Directory:**
   - We use `os.makedirs('output', exist_ok=True)` to create an `output/` folder if it does not already exist. The `exist_ok=True` parameter prevents errors if the directory already exists.

2. **Exporting Grouped Summary Data:**
   - We export the `species_summary` DataFrame to `output/species_summary.csv` using the `.to_csv()` function, preserving the species names as index labels.

3. **Exporting Filtered Data:**
   - We save the `filtered_sepal` DataFrame to `output/filtered_iris.csv`. The `index=False` parameter is used to avoid writing row index numbers into the file.

4. **Confirmation:**
   - A confirmation message is printed to verify that all files were exported successfully.
